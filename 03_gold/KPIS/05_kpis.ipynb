```python
# Load Tables

from pyspark.sql.functions import col, avg

fact_df = spark.table("medical_project.gold.fact_encounters")
payer_df = spark.table("medical_project.gold.dim_payer")

display(fact_df)
```

```python
# Aggregate Avg Cost

agg_df = fact_df.groupBy("payer_id").agg(
    avg("total_cost").alias("avg_claim_cost")
)
```

```python
# Join with Payer Dimension

kpi5 = agg_df.join(
    payer_df,
    "payer_id",
    "left"
)
```

```python
# Final Output

kpi5 = kpi5.select(
    "payer_id",
    "payer_name",   
    "avg_claim_cost"
).orderBy(col("avg_claim_cost").desc())

display(kpi5)
```

```python
# Save Table 
kpi5.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("medical_project.gold.kpi_avg_claim_cost_by_payer")
```
